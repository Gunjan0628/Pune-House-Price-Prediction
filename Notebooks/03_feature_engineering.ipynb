# 03. Feature Engineering

## 1. Introduction

Feature engineering is the process of selecting, transforming, and preparing relevant input variables for a machine learning model.

In this notebook, we will prepare the cleaned Pune property dataset for house price prediction. We will inspect the available features, select suitable input variables, identify missing values, and investigate unusual records.

The cleaned dataset generated in the data cleaning notebook will be used as the input.

### Objectives
- Load the cleaned dataset.
- Understand the available columns and data types.
- Define the target variable and input features.
- Review missing values and unusual records.
- Prepare the feature set for model training.

**Target variable:** `price`

**Problem type:** Supervised learning — Regression

```python
import pandas as pd
import numpy as np
from pathlib import Path

pd.set_option("display.max_columns", None)
pd.set_option("display.float_format", "{:,.2f}".format)

# Locate the project root
project_root = Path.cwd().parent

# Load the cleaned dataset
cleaned_file = (
    project_root
    / "data"
    / "processed"
    / "pune_property_cleaned.csv"
)

df = pd.read_csv(cleaned_file)

print("Dataset loaded successfully!")
print("Dataset shape:", df.shape)

display(df.head())
```

```python
print("Number of rows:", df.shape[0])
print("Number of columns:", df.shape[1])

print("\nColumn names:")
print(df.columns.tolist())

print("\nData types:")
display(df.dtypes.to_frame(name="data_type"))
```

```python
# Target variable
y = df["price"].copy()

# Initial input features
features = [
    "area",
    "bhk_count",
    "property_type",
    "locality",
    "bathroom",
    "balconies",
    "floor_number",
    "totalfloor",
    "property_age_years",
    "facing",
    "ownership",
    "neworold",
    "status"
]

X = df[features].copy()

print("Feature dataset shape:", X.shape)
print("Target dataset shape:", y.shape)

print("\nSelected features:")
print(X.columns.tolist())

display(X.head())
```

```python
missing_summary = pd.DataFrame({
    "missing_count": X.isna().sum(),
    "missing_percentage": (X.isna().mean() * 100).round(2)
}).sort_values("missing_percentage", ascending=False)

display(missing_summary)
```

```python
print("Missing prices:", y.isna().sum())
print("Zero or negative prices:", (y <= 0).sum())

print("\nPrice summary:")
display(y.describe())

print("\nNumber of unique prices:", y.nunique())
```

```python
print("Records with BHK count equal to zero:")

display(
    df.loc[
        df["bhk_count"].eq(0),
        ["price", "area", "bhk", "property_type", "locality"]
    ]
)

print("Records with missing BHK count:", df["bhk_count"].isna().sum())
```

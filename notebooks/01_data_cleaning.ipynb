# Pune House Price Prediction — Data Cleaning

## 1. Project Objective
The objective of this notebook is to inspect, clean, and validate the raw Pune property dataset before performing exploratory data analysis and developing machine learning models.

## 2. Data Cleaning Workflow
- Inspect the dataset structure and dimensions.
- Identify missing values and duplicate records.
- Examine data types and inconsistent values.
- Investigate invalid records and potential outliers.
- Apply appropriate cleaning techniques.
- Validate and save the cleaned dataset.

**Important:** The original raw dataset will remain unchanged. All cleaning operations will be performed on a separate working copy.


```python
import pandas as pd
import numpy as np
import re
from pathlib import Path
from IPython.display import display
```

```python
PROJECT_ROOT = Path(r"D:/Pune-House-Price-Prediction")

RAW_PATH = PROJECT_ROOT / "data" / "raw" / "Pune_property_data.csv"

OUTPUT_PATH = PROJECT_ROOT / "data" / "processed" / "pune_cleaned_reference.csv"

OUTPUT_PATH.parent.mkdir(parents=True, exist_ok=True)
```

```python
df = pd.read_csv(RAW_PATH)

print("Dataset loaded successfully!")
print("Rows and columns:", df.shape)

display(df.head())
```

```python
print("Column names:")
print(df.columns.tolist())

print("\nData types:")
display(df.dtypes.to_frame(name="Data Type"))

print("\nDataset information:")
df.info()
```

```python
cleaned_df = df.copy()

print("Working copy created.")
print("Shape:", cleaned_df.shape)
```

```python
header_mask = (
    cleaned_df["neworold"].astype("string").str.strip().str.lower().eq("neworold")
    | cleaned_df["ownership"].astype("string").str.strip().str.lower().eq("ownership")
    | cleaned_df["status"].astype("string").str.strip().str.lower().eq("status")
)

print("Rows identified as repeated headers:", int(header_mask.sum()))

cleaned_df = cleaned_df.loc[~header_mask].copy()

print("Shape after removing repeated headers:", cleaned_df.shape)
```

```python
text_columns = cleaned_df.select_dtypes(
    include=["object", "string"]
).columns

for column in text_columns:
    cleaned_df[column] = cleaned_df[column].astype("string").str.strip()

print("Text columns standardized.")
```

```python
lowercase_columns = ["facing", "locality", "neworold"]

for column in lowercase_columns:
    cleaned_df[column] = cleaned_df[column].str.lower()

print("Selected categorical columns converted to lowercase.")

display(cleaned_df[lowercase_columns].head())
```

```python
cleaned_df["ownership"] = cleaned_df["ownership"].replace({
    "Leasehol": "leasehold",
    "leasehol": "leasehold",
    "LEASEHOL": "leasehold"
})

print(cleaned_df["ownership"].value_counts(dropna=False).head(10))
```

```python
cleaned_df["Area"] = pd.to_numeric(
    cleaned_df["area"]
    .astype("string")
    .str.replace(",", "", regex=False)
    .str.extract(r"(\d+(?:\.\d+)?)", expand=False),
    errors="coerce"
)

display(cleaned_df[["area", "Area"]].head(10))
```

```python
age_text = cleaned_df["age"].astype("string").str.strip()

age_range_upper = pd.to_numeric(
    age_text.str.extract(r"-\s*(\d+(?:\.\d+)?)", expand=False),
    errors="coerce"
)

age_single_value = pd.to_numeric(
    age_text.str.extract(r"^\s*(\d+(?:\.\d+)?)\s*years?\s*$", expand=False),
    errors="coerce"
)

cleaned_df["Age"] = age_range_upper.fillna(age_single_value)

display(cleaned_df[["age", "Age"]].head(15))
```

```python
cleaned_df["addi_rooms"] = pd.to_numeric(
    cleaned_df["additionalrooms"]
    .astype("string")
    .str.extract(r"^\s*(\d+)\s*rooms?", expand=False),
    errors="coerce"
)

display(cleaned_df[["additionalrooms", "addi_rooms"]].head(15))
```

```python
numeric_columns = ["balconies", "bathroom", "opensides"]

for column in numeric_columns:
    cleaned_df[column] = pd.to_numeric(
        cleaned_df[column],
        errors="coerce"
    )

display(cleaned_df[numeric_columns].dtypes.to_frame(name="Data Type"))
```

```python
cleaned_df["floor"] = pd.to_numeric(
    cleaned_df["floor"]
    .astype("string")
    .str.extract(r"(-?\d+(?:\.\d+)?)", expand=False),
    errors="coerce"
)

display(cleaned_df[["floor"]].head(15))
```

```python
bhk_text = cleaned_df["bhk"].astype("string").str.strip()

bedrooms_bhk = pd.to_numeric(
    bhk_text.str.extract(r"(\d+)\s*BHK", expand=False),
    errors="coerce"
)

bedrooms_rk = pd.to_numeric(
    bhk_text.str.extract(r"(\d+)\s*RK", expand=False),
    errors="coerce"
)

cleaned_df["Bedrooms"] = bedrooms_bhk.fillna(bedrooms_rk)

display(cleaned_df[["bhk", "Bedrooms"]].head(20))
```

```python
bhk_lower = cleaned_df["bhk"].astype("string").str.lower()

cleaned_df["Residential"] = np.where(
    bhk_lower.str.contains("residential plot", na=False),
    "yes",
    "no"
)

display(cleaned_df[["bhk", "Residential"]].head(15))
```

```python
cleaned_df["Villa"] = np.where(
    bhk_lower.str.contains(r"\bvilla\b", na=False, regex=True),
    "yes",
    "no"
)

display(cleaned_df[["bhk", "Villa"]].head(15))
```

```python
cleaned_df["Independent"] = np.where(
    bhk_lower.str.contains("independent", na=False),
    "yes",
    "no"
)

display(cleaned_df[["bhk", "Independent"]].head(15))
```

```python
cleaned_df["Penthouse"] = np.where(
    bhk_lower.str.contains("penthouse", na=False),
    "yes",
    "no"
)

display(cleaned_df[["bhk", "Penthouse"]].head(15))
```

```python
def create_apartment_description(bedrooms):
    if pd.isna(bedrooms):
        return np.nan

    bedrooms = int(bedrooms)

    bedroom_label = "Bedroom" if bedrooms == 1 else "Bedrooms"

    return f"{bedrooms} {bedroom_label}, 1 Hall, 1 Kitchen"


cleaned_df["Apartment"] = cleaned_df["Bedrooms"].apply(
    create_apartment_description
)

display(cleaned_df[["bhk", "Apartment", "Bedrooms"]].head(20))
```

```python
amenity_columns = [
    "Lift Available",
    "Car Parking",
    "Power Backup",
    "24 X 7 Security",
    "Children play area",
    "Vaastu Compliant",
    "Club House",
    "Gymnasium",
    "Swimming Pool",
    "Sports Facility",
    "Indoor Games",
    "Jogging Track",
    "Maintenance Staff",
    "Intercom",
    "Golf Course",
    "Cafeteria",
    "Rain Water Harvesting",
    "Staff Quarter",
    "Multipurpose Room",
    "Landscaped Gardens",
    "Shopping Mall",
    "School",
    "Hospital",
    "ATM",
    "AC",
    "Wardrobe",
    "TV",
    "Refrigerator",
    "Sofa",
    "Washing Machine",
    "Wifi",
    "Microwave",
    "Dining Table",
    "Gas connection",
    "BED"
]

print("Total amenity columns:", len(amenity_columns))
```

```python

amenities_text = cleaned_df["amenitiesavailable"].astype("string")

for amenity in amenity_columns:

    found = amenities_text.str.contains(
        re.escape(amenity),
        case=False,
        regex=True,
        na=False
    )

    # Use pandas StringDtype to support strings and missing values
    cleaned_df[amenity] = pd.Series(
        np.select(
            [
                amenities_text.isna(),
                found
            ],
            [
                pd.NA,
                "yes"
            ],
            default="no"
        ),
        index=cleaned_df.index,
        dtype="string"
    )

display(cleaned_df[amenity_columns].head())
```

```python
reference_columns = [
    "Unnamed: 0",
    "balconies",
    "bathroom",
    "facing",
    "floor",
    "locality",
    "neworold",
    "opensides",
    "price",
    "addi_rooms",
    "Age",
    "Lift Available",
    "Car Parking",
    "Power Backup",
    "24 X 7 Security",
    "Children play area",
    "Vaastu Compliant",
    "Club House",
    "Gymnasium",
    "Swimming Pool",
    "Sports Facility",
    "Indoor Games",
    "Jogging Track",
    "Maintenance Staff",
    "Intercom",
    "Golf Course",
    "Cafeteria",
    "Rain Water Harvesting",
    "Staff Quarter",
    "Multipurpose Room",
    "Landscaped Gardens",
    "Shopping Mall",
    "School",
    "Hospital",
    "ATM",
    "AC",
    "Wardrobe",
    "TV",
    "Refrigerator",
    "Sofa",
    "Washing Machine",
    "Wifi",
    "Microwave",
    "Dining Table",
    "Gas connection",
    "BED",
    "Area",
    "Residential",
    "Villa",
    "Independent",
    "Penthouse",
    "Apartment",
    "Bedrooms"
]

final_df = cleaned_df[reference_columns].copy()

print("Final dataset shape:", final_df.shape)
print("Number of columns:", len(final_df.columns))

display(final_df.head())
```

```python
print("Final columns:")

for number, column in enumerate(final_df.columns, start=1):
    print(number, column)

assert len(final_df.columns) == 53
assert final_df.columns.tolist() == reference_columns

print("\nAll 53 columns are present in the required order.")
```

```python
final_df.to_csv(OUTPUT_PATH, index=False)

print("Cleaned dataset saved successfully!")
print("File location:", OUTPUT_PATH)
print("Final shape:", final_df.shape)
```

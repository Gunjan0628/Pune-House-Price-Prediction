```python
price_series = df["price"].dropna()

Q1 = price_series.quantile(0.25)
Q3 = price_series.quantile(0.75)
IQR = Q3 - Q1

lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR

price_outliers = df[
    (df["price"] < lower_bound) |
    (df["price"] > upper_bound)
]

print("Lower IQR bound:", lower_bound)
print("Upper IQR bound:", upper_bound)
print("Potential price outliers:", len(price_outliers))

display(price_outliers.head(10))
```

```python
amenity_columns = [
    "Lift Available",
    "Car Parking",
    "Power Backup",
    "24 X 7 Security",
    "Children play area",
    "Vaastu Compliant",
    "Club House",
    "Gymnasium",
    "Swimming Pool",
    "Sports Facility",
    "Indoor Games",
    "Jogging Track",
    "Maintenance Staff",
    "Intercom",
    "Golf Course",
    "Cafeteria",
    "Rain Water Harvesting",
    "Staff Quarter",
    "Multipurpose Room",
    "Landscaped Gardens",
    "Shopping Mall",
    "School",
    "Hospital",
    "ATM",
    "AC",
    "Wardrobe",
    "TV",
    "Refrigerator",
    "Sofa",
    "Washing Machine",
    "Wifi",
    "Microwave",
    "Dining Table",
    "Gas connection",
    "BED"
]

amenity_columns = [
    column for column in amenity_columns
    if column in df.columns
]

print("Amenities found:", len(amenity_columns))
print(amenity_columns)
```

```python
for column in amenity_columns:
    print(f"\n{column}")
    display(df[column].value_counts(dropna=False).to_frame("Count"))
```

```python
amenity_summary = []

for column in amenity_columns:
    values = df[column].astype("string").str.strip().str.lower()

    yes_count = values.eq("yes").sum()
    no_count = values.eq("no").sum()

    amenity_summary.append({
        "Amenity": column,
        "Yes Count": int(yes_count),
        "No Count": int(no_count),
        "Missing Count": int(values.isna().sum()),
        "Yes Percentage": round(
            yes_count / len(df) * 100, 2
        ) if len(df) else 0
    })

amenity_summary = pd.DataFrame(amenity_summary)
amenity_summary = amenity_summary.sort_values(
    "Yes Count", ascending=False
)

display(amenity_summary)
```

```python
top_amenities = amenity_summary.head(20).sort_values("Yes Count")

plt.figure(figsize=(11, 8))

plt.barh(
    top_amenities["Amenity"],
    top_amenities["Yes Count"]
)

plt.title("Top 20 Amenities by Number of Properties")
plt.xlabel("Properties marked yes")
plt.ylabel("Amenity")
plt.tight_layout()
plt.show()
```

```python
def get_selected_amenities(row):
    selected = []

    for column in amenity_columns:
        value = str(row[column]).strip().lower()

        if value == "yes":
            selected.append(column)

    return selected


df["selected_amenities"] = df.apply(
    get_selected_amenities,
    axis=1
)

display(
    df[["selected_amenities"] + amenity_columns[:5]].head(10)
)
```

```python
df["amenity_count"] = df["selected_amenities"].apply(len)

display(df["amenity_count"].describe().to_frame("Amenity Count Statistics"))

plt.figure(figsize=(9, 5))

sns.histplot(
    data=df,
    x="amenity_count",
    bins=20,
    kde=True
)

plt.title("Number of Listed Amenities per Property")
plt.xlabel("Number of amenities marked yes")
plt.ylabel("Number of properties")
plt.tight_layout()
plt.show()
```

```python
TARGET = "price"

identifier_columns = [
    column for column in ["Unnamed: 0"]
    if column in df.columns
]

amenity_columns = [
    column for column in amenity_columns
    if column in df.columns
]

candidate_features = [
    column for column in df.columns
    if column not in identifier_columns
    and column not in [TARGET, "selected_amenities", "amenity_count"]
]

print("Target:", TARGET)
print("Identifier columns excluded:", identifier_columns)
print("Amenity columns:", len(amenity_columns))
print("Candidate features:", len(candidate_features))

display(pd.DataFrame({
    "Candidate Feature": candidate_features,
    "Data Type": [str(df[col].dtype) for col in candidate_features],
    "Missing Values": [df[col].isna().sum() for col in candidate_features],
    "Unique Values": [df[col].nunique(dropna=True) for col in candidate_features]
}))
```

```python
EDA_OUTPUT_DIR = PROJECT_ROOT / "reports" / "eda"
EDA_OUTPUT_DIR.mkdir(parents=True, exist_ok=True)

missing_summary.to_csv(
    EDA_OUTPUT_DIR / "missing_values_summary.csv"
)

amenity_summary.to_csv(
    EDA_OUTPUT_DIR / "amenity_summary.csv",
    index=False
)

if "price" in correlation_matrix.columns:
    price_correlations.to_csv(
        EDA_OUTPUT_DIR / "price_correlations.csv"
    )

print("EDA summary files saved to:", EDA_OUTPUT_DIR)
```

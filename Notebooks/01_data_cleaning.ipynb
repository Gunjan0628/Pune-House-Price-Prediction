3.1 Identify Invalid and Repeated Header Rows

```python
suspect_columns = ["neworold", "status", "ownership"]

suspect_masks = {}

for column in suspect_columns:
    suspect_masks[column] = (
        df[column]
        .astype("string")
        .str.strip()
        .str.casefold()
        .eq(column.casefold())
        .fillna(False)
    )

    print(
        f"{column}:",
        int(suspect_masks[column].sum()),
        "suspicious records"
    )

combined_suspect_mask = pd.DataFrame(suspect_masks).any(axis=1)

print("\nTotal records matching at least one suspicious label:",
      int(combined_suspect_mask.sum()))

print("\nSuspicious records:")
display(df.loc[combined_suspect_mask, suspect_columns].head(30))
```

```python
display(df.loc[combined_suspect_mask].head(30))
```

```python
df_clean = df.copy(deep=True)

# Remove the sequential exported row index
df_clean = df_clean.drop(columns=["Unnamed: 0"])

# Remove leading/trailing whitespace from text values
text_columns = df_clean.select_dtypes(
    include=["object", "str", "string"]
).columns

for column in text_columns:
    df_clean[column] = df_clean[column].str.strip()

print("Shape after removing index column:", df_clean.shape)
```

```python
missing_markers = [
    "",
    "-",
    "NA",
    "N/A",
    "NaN",
    "None",
    "null",
    "Not Available",
    "Not available",
]

for column in df_clean.select_dtypes(
    include=["object", "str", "string"]
).columns:
    values = df_clean[column].astype("string").str.strip()

    df_clean[column] = values.mask(
        values.isin(missing_markers),
        pd.NA
    )

print("Blank and recognized placeholder values standardized.")
```

3.2 Clean Numeric Features

```python
# Price: values such as "27000000"
df_clean["price"] = pd.to_numeric(
    df_clean["price"].str.replace(",", "", regex=False),
    errors="coerce"
)

# Area: values such as "1,611 sq ft"
df_clean["area"] = pd.to_numeric(
    df_clean["area"]
    .str.replace(",", "", regex=False)
    .str.extract(r"(\d+(?:\.\d+)?)", expand=False),
    errors="coerce"
)

# Carpet area: values such as "1209 sq ft"
df_clean["carpetarea"] = pd.to_numeric(
    df_clean["carpetarea"]
    .str.replace(",", "", regex=False)
    .str.extract(r"(\d+(?:\.\d+)?)", expand=False),
    errors="coerce"
)

# Price per square foot: values such as "₹ 16,759/ sq ft"
df_clean["pricepersquare"] = pd.to_numeric(
    df_clean["pricepersquare"]
    .str.replace(",", "", regex=False)
    .str.extract(r"(\d+(?:\.\d+)?)", expand=False),
    errors="coerce"
)

print(df_clean[[
    "price",
    "pricepersquare",
    "area",
    "carpetarea"
]].dtypes)

display(df_clean[[
    "price",
    "pricepersquare",
    "area",
    "carpetarea"
]].head(10))
```

```python
bhk_text = df_clean["bhk"].astype("string").str.strip()

df_clean["bhk_count"] = pd.to_numeric(
    bhk_text.str.extract(r"(\d+)\s*(?:BHK|RK)\b", flags=2, expand=False),
    errors="coerce"
)

df_clean["property_type"] = (
    bhk_text
    .str.replace(r"^\s*\d+\s*(?:BHK|RK)\s*", "", regex=True, case=False)
    .str.strip()
)

df_clean.loc[
    df_clean["property_type"].eq(""),
    "property_type"
] = pd.NA

display(df_clean[["bhk", "bhk_count", "property_type"]].head(20))
```

### 3.5 Standardize Floor Information

The floor column contains inconsistent formatting, including trailing commas and the abbreviation "Gr". A separate numeric feature, `floor_number`, is created to standardize floor values while preserving the original column for reference. "Gr" is interpreted as ground floor (0).

```python
# Normalize text formatting
floor_text = (
    df_clean["floor"]
    .astype("string")
    .str.strip()
)

# Remove trailing commas and surrounding spaces
floor_text = floor_text.str.replace(
    r",+$", "", regex=True
).str.strip()

# Create a separate numeric feature
df_clean["floor_number"] = pd.to_numeric(
    floor_text,
    errors="coerce"
).astype("Int64")

# Interpret "Gr" as ground floor (0)
ground_mask = floor_text.str.casefold().eq("gr").fillna(False)
df_clean.loc[ground_mask, "floor_number"] = 0

# Validate the conversion
print("Original floor column:")
display(df_clean["floor"].value_counts(dropna=False).head(15))

print("\nNew floor_number column:")
display(df_clean["floor_number"].value_counts(dropna=False).sort_index())

print("\nFloor values that could not be converted:")
unconverted_mask = (
    floor_text.notna()
    & df_clean["floor_number"].isna()
)
display(
    df_clean.loc[unconverted_mask, ["floor"]]
    .drop_duplicates()
)
```

```python
invalid_floor_mask = (
    df_clean["floor_number"].notna()
    & df_clean["totalfloor"].notna()
    & (df_clean["floor_number"] > df_clean["totalfloor"])
)

print("Rows where floor exceeds total floors:",
      int(invalid_floor_mask.sum()))

display(
    df_clean.loc[
        invalid_floor_mask,
        ["floor", "floor_number", "totalfloor", "locality"]
    ].head(20)
)
```

### 3.6 Extract Road Width

The road-facing width column contains measurements stored as text, such as "20 feet". A separate numeric feature, `road_width_feet`, is created to support quantitative analysis. Missing values are preserved.

```python
# Keep the original roadfaceing column unchanged
road_text = (
    df_clean["roadfaceing"]
    .astype("string")
    .str.strip()
)

# Extract the numeric part of the measurement
df_clean["road_width_feet"] = pd.to_numeric(
    road_text.str.extract(
        r"(\d+(?:\.\d+)?)",
        expand=False
    ),
    errors="coerce"
).astype("Float64")

# Inspect the result
print("Original road-facing values:")
display(df_clean["roadfaceing"].value_counts().head(15))

print("\nCleaned road width summary:")
display(df_clean["road_width_feet"].describe())

print("\nUnique numeric road widths:")
display(
    df_clean["road_width_feet"]
    .value_counts()
    .sort_index()
    .head(30)
)

print(
    "\nMissing road widths:",
    int(df_clean["road_width_feet"].isna().sum())
)
```

### 3.7 Extract Property Age

The age column contains ranges and single-year descriptions. A numeric feature, `property_age_years`, is created using the midpoint for ranges and the stated value for single-year entries. The original age column is retained.

```python
age_text = (
    df_clean["age"]
    .astype("string")
    .str.strip()
    .str.casefold()
)

# Extract the first and second numbers
age_numbers = age_text.str.extract(
    r"(\d+(?:\.\d+)?)\s*(?:-\s*(\d+(?:\.\d+)?))?",
    expand=True
)

age_start = pd.to_numeric(age_numbers[0], errors="coerce")
age_end = pd.to_numeric(age_numbers[1], errors="coerce")

# Use the midpoint for ranges; use the stated value otherwise
df_clean["property_age_years"] = (
    age_start + age_end.fillna(age_start)
) / 2

df_clean["property_age_years"] = (
    df_clean["property_age_years"].astype("Float64")
)

print("Original age examples:")
display(df_clean["age"].dropna().drop_duplicates().head(20))

print("\nConverted property age:")
display(
    df_clean[["age", "property_age_years"]]
    .drop_duplicates()
    .head(30)
)

print("\nProperty age summary:")
display(df_clean["property_age_years"].describe())
```

```python
df_clean["possession_date"] = pd.to_datetime(
    df_clean["possesiondate"],
    format="%d-%m-%Y",
    errors="coerce"
)

print("Possession date data type:",
      df_clean["possession_date"].dtype)

print("Missing possession dates:",
      int(df_clean["possession_date"].isna().sum()))

print("\nEarliest possession date:",
      df_clean["possession_date"].min())

print("Latest possession date:",
      df_clean["possession_date"].max())

display(
    df_clean[["possesiondate", "possession_date"]]
    .dropna()
    .head(10)
)
```

## Data Cleaning Summary

The original dataset contained 37,539 rows and 24 columns.

The following cleaning and preprocessing steps were performed:

1. Removed the unnecessary sequential index column.
2. Trimmed whitespace and standardized common missing-value markers.
3. Identified and removed 22 repeated header rows.
4. Converted price, area, carpet area, and price per square unit into numeric formats.
5. Created `bhk_count` and `property_type` from the original BHK field.
6. Created numeric features for floor number, road width, and property age.
7. Converted the possession date into a datetime format.
8. Standardized the ownership value `Leasehol` to `Leasehold`.
9. Removed one completely empty row.
10. Removed 7,836 exact duplicate rows.

**Final dataset:** 29,680 rows and 29 columns.

**Final validation:** No fully empty rows, no exact duplicate rows, no missing prices, and no non-positive prices remain.

Missing values in other features were retained for further investigation and appropriate handling during model preprocessing.

Potential anomalies, including unusually large road-width values and the `0 BHK Villa` record, should be reviewed before final model training.
